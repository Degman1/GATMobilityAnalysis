### Compute Node-Wise Thresholded-WMAPE

```python
cols = dataset.data_pd.columns.astype(int).tolist()

y_pred_cleaned = y_pred.reshape((y_pred.shape[0] * y_pred.shape[1], -1))[:config["N_NODES"] * len(d_test), :]
y_truth_cleaned = y_truth.reshape((y_truth.shape[0] * y_truth.shape[1], -1))[:config["N_NODES"] * len(d_test), :]

# Reshape to [num_graphs, num_nodes, pred_len]
y_pred_cleaned = y_pred_cleaned.reshape(len(d_test), config["N_NODES"], -1)
y_truth_cleaned = y_truth_cleaned.reshape(len(d_test), config["N_NODES"], -1)

# Calculate WMAPE per node at 3, 6, and 9 prediction steps
node_wmape_scores = []

for node_idx in range(config["N_NODES"]):
    
    sum_score_3 = 0
    sum_score_6 = 0
    sum_score_9 = 0
    n_score_3 = 0
    n_score_6 = 0
    n_score_9 = 0
    
    for time_idx in range(len(d_test)):
        pred_node = y_pred_cleaned[time_idx, node_idx, :]
        truth_node = y_truth_cleaned[time_idx, node_idx, :]
        
        s3 = utils.math.WMAPE(pred_node[:3], truth_node[:3])
        s6 = utils.math.WMAPE(pred_node[:6], truth_node[:6])
        s9 = utils.math.WMAPE(pred_node, truth_node)
        
        if not torch.isnan(s3):
            sum_score_3 += s3
            n_score_3 += 1
        if not torch.isnan(s6):
            sum_score_6 += s6
            n_score_6 += 1
        if not torch.isnan(s9):
            sum_score_9 += s9
            n_score_9 += 1
            
    node_wmape_scores.append((cols[node_idx], sum_score_3 / n_score_3 , sum_score_6 / n_score_6, sum_score_9 / n_score_9))

# Sort nodes by full 9-step WMAPE
sorted_nodes = sorted(node_wmape_scores, key=lambda x: x[3])

print(f"WMAPE Average Full 3-step: {np.mean([x[1] for x in node_wmape_scores])}")
print(f"WMAPE Average Full 6-step: {np.mean([x[2] for x in node_wmape_scores])}")
print(f"WMAPE Average Full 9-step: {np.mean([x[3] for x in node_wmape_scores])}")

# Print top 50 nodes with best (lowest) 9-step WMAPE
print("Top 10 nodes by lowest 9-step WMAPE:")
print(f"{'Node':<8}{'WMAPE@3':>10}{'WMAPE@6':>10}{'WMAPE@9':>10}")
for node, s3, s6, s9 in sorted_nodes[:10]:
    print(f"{node:<8}{s3:>10.2f}{s6:>10.2f}{s9:>10.2f}")

# Sort nodes by 9-step WMAPE
sorted_nodes = sorted(node_wmape_scores, key=lambda x: x[3])  # x[3] is 9-step WMAPE

# Extract WMAPE values for each step
wmape_3_sorted = [x[1] for x in sorted_nodes]
wmape_6_sorted = [x[2] for x in sorted_nodes]
wmape_9_sorted = [x[3] for x in sorted_nodes]

to_remove = [x[0] for x in sorted_nodes[250:]]
print(f"Remove these nodes manually?: {to_remove}")

# Plot all three curves
plt.figure(figsize=(10, 6))
plt.plot(wmape_3_sorted, label='3-Step WMAPE', marker='o', markersize=2, linewidth=1)
plt.plot(wmape_6_sorted, label='6-Step WMAPE', marker='o', markersize=2, linewidth=1)
plt.plot(wmape_9_sorted, label='9-Step WMAPE', marker='o', markersize=2, linewidth=1)

plt.xlabel("Node Rank (Sorted by Best 9-Step WMAPE)")
plt.ylabel("WMAPE (%)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
location = "./output/WMAPE_by_node.png"
plt.savefig(location, bbox_inches="tight")
```

```python
wmape_3_sorted_t = torch.tensor(wmape_3_sorted)
wmape_6_sorted_t = torch.tensor(wmape_6_sorted)
wmape_9_sorted_t = torch.tensor(wmape_9_sorted)

# Compute Q3 and IQR for wmape9
q3 = torch.quantile(wmape_9_sorted_t, 0.75)
iqr = q3 - torch.quantile(wmape_9_sorted_t, 0.25)
upper_bound = q3 + 1.5 * iqr

# Only exclude high outliers
non_outlier_mask = wmape_9_sorted_t <= upper_bound
non_outlier_indices = torch.nonzero(non_outlier_mask, as_tuple=False).flatten().tolist()

# Compute averages over non-outliers
avg_wmape3_all = wmape_3_sorted_t[non_outlier_indices].mean().item()
avg_wmape6_all = wmape_6_sorted_t[non_outlier_indices].mean().item()
avg_wmape9_all = wmape_9_sorted_t[non_outlier_indices].mean().item()

print("\nAll Non-Outlier Nodes (n={}):".format(len(non_outlier_indices)))
print(f"  Avg 3-step WMAPE: {avg_wmape3_all:.2f}%")
print(f"  Avg 6-step WMAPE: {avg_wmape6_all:.2f}%")
print(f"  Avg 9-step WMAPE: {avg_wmape9_all:.2f}%")

# NOTE This plot uses kernel density estimation (KDE) with Seaborn

# extract the non-outlier arrays and convert to numpy
w3 = wmape_3_sorted_t[non_outlier_mask].cpu().numpy()
w6 = wmape_6_sorted_t[non_outlier_mask].cpu().numpy()
w9 = wmape_9_sorted_t[non_outlier_mask].cpu().numpy()

# Avoid negative densities and set consistent bandwidth for stability
plt.figure(figsize=(10, 6))
sns.kdeplot(w3, label='3-Step Threshold-WMAPE', bw_adjust=0.8, clip=(0, None))
sns.kdeplot(w6, label='6-Step Threshold-WMAPE', bw_adjust=0.8, clip=(0, None))
sns.kdeplot(w9, label='9-Step Threshold-WMAPE', bw_adjust=0.8, clip=(0, None))

plt.xlabel("Windowed WMAPE (%)")
plt.ylabel("Density")
# plt.title("Windowed WMAPE Distribution by Forecast Horizon")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("./output/wmape_density.png", bbox_inches="tight")
```

### Compute Node-Wise Windowed-Thresholded-WMAPE

```python
cols = dataset.data_pd.columns.astype(int).tolist()

y_pred_cleaned = y_pred.reshape((y_pred.shape[0] * y_pred.shape[1], -1))[:config["N_NODES"] * len(d_test), :]
y_truth_cleaned = y_truth.reshape((y_truth.shape[0] * y_truth.shape[1], -1))[:config["N_NODES"] * len(d_test), :]

# Reshape to [num_graphs, num_nodes, pred_len]
y_pred_cleaned = y_pred_cleaned.reshape(len(d_test), config["N_NODES"], -1)
y_truth_cleaned = y_truth_cleaned.reshape(len(d_test), config["N_NODES"], -1)

w = 2   # Window radius
threshold = 0.01    # min threshold for ground truth value

wwmape3 = torch.zeros(config["N_NODES"])
wwmape6 = torch.zeros(config["N_NODES"])
wwmape9 = torch.zeros(config["N_NODES"])
count3 = torch.zeros(config["N_NODES"])
count6 = torch.zeros(config["N_NODES"])
count9 = torch.zeros(config["N_NODES"])

# Only want my test data ground truths padded by the window size on the left side
start_gt = len(d_train) + len(d_val) + config["N_HIST"] - w
full_ground_truth = torch.tensor(dataset.data_pd.values)[start_gt:, :]
# print(full_ground_truth.shape)

# NOTE This could be vectorized to run significantly faster, but I unfortunately don't have the time to convert it right now.
for node_idx in range(config["N_NODES"]):
    
    T = full_ground_truth.size(0)
    gt = full_ground_truth[:, node_idx]
    last  = gt[-1].expand(w)
    padded = torch.cat([gt, last], dim=0)
    windows = padded.unfold(0, 2*w+1, 1)
    
    for time_idx in range(0, len(d_test)):
        predictions = y_pred_cleaned[time_idx, node_idx, :]
        ground_truths = y_truth_cleaned[time_idx, node_idx, :]
        indices = torch.arange(config["N_PRED"]) + time_idx
        ground_truth_windows = windows[indices]
        abs_errors = torch.abs(predictions.unsqueeze(1) - ground_truth_windows)
        min_errors = abs_errors.min(dim=1).values
        denoms = torch.abs(ground_truths)
        mask = torch.abs(ground_truths) >= threshold
        
        if mask[:3].any():
            num = torch.sum(min_errors[:3][mask[:3]])
            den = torch.sum(denoms[:3][mask[:3]])
            wwmape3[node_idx] += (num / den) * 100
            count3[node_idx] += 1
        
        if mask[:6].any():
            num = torch.sum(min_errors[:6][mask[:6]])
            den = torch.sum(denoms[:6][mask[:6]])
            wwmape6[node_idx] += (num / den) * 100
            count6[node_idx] += 1
        
        if mask.any():
            num = torch.sum(min_errors[mask])
            den = torch.sum(denoms[mask])
            wwmape9[node_idx] += (num / den) * 100
            count9[node_idx] += 1

wwmape3 /= count3 + 1e-6  # To avoid division by zero
wwmape6 /= count6 + 1e-6  # To avoid division by zero
wwmape9 /= count9 + 1e-6  # To avoid division by zero
print(wwmape9)
```

```python
# Sort and display top nodes
for h, (scores, counts) in zip([3,6,9], [(wwmape3, count3), (wwmape6, count6), (wwmape9, count9)]):
    sorted_scores, indices = torch.sort(scores)
    print(f"\nTop 10 nodes by lowest {h}-step WMAPE:")
    for i in indices[:10]:
        print(f"  Node {cols[i]}: {scores[i]:.2f}% (n={int(counts[i])})")

_, idx9_sorted = torch.sort(wwmape9)
wwmape_3_sorted = wwmape3[idx9_sorted].tolist()
wwmape_6_sorted = wwmape6[idx9_sorted].tolist()
wwmape_9_sorted = wwmape9[idx9_sorted].tolist()

# Plot all three curves
plt.figure(figsize=(10, 6))
plt.plot(wwmape_3_sorted, label='3-Step Windowed-WMAPE', marker='o', markersize=2, linewidth=1)
plt.plot(wwmape_6_sorted, label='6-Step Windowed-WMAPE', marker='o', markersize=2, linewidth=1)
plt.plot(wwmape_9_sorted, label='9-Step Windowed-WMAPE', marker='o', markersize=2, linewidth=1)
plt.plot(count9[idx9_sorted], label='Number of Prediction Windows Contributing To Scores (Due to Threshold)', marker='o', markersize=2, linewidth=1)

avg_sorted_3 = torch.tensor(wwmape_3_sorted).mean().item()
avg_sorted_6 = torch.tensor(wwmape_6_sorted).mean().item()
avg_sorted_9 = torch.tensor(wwmape_9_sorted).mean().item()

plt.xlabel("Node Rank (Sorted by Best 9-Step WMAPE)")
# plt.ylabel("WMAPE (%)")
plt.legend()
plt.grid(True, alpha=0.6)
plt.tight_layout()
location = "./output/Windowed-WMAPE_by_node.png"
plt.savefig(location, bbox_inches="tight")
```

```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display
from IPython.display import Markdown
from IPython.display import clear_output
import json

s = y_truth.shape
y_truth_reshaped = y_truth.reshape(s[0], config["BATCH_SIZE"], config["N_NODES"], s[-1])
plots = []
for j in range(len(final_sorted_points)):
    i = final_sorted_indices[j]
    # just get the first prediction out for the nth node
    plot = y_truth_reshaped[:, :, i, 0]
    # Flatten to get the predictions for entire test dataset
    plot = torch.flatten(plot)
    plots.append(plot)

# Per-plot clicked x positions
clicked_x_coords = [set() for _ in range(len(plots))]

# Load marks from JSON file if it exists
def load_marks_from_file(filename="anomaly_labels.json"):
    try:
        with open(filename, 'r') as f:
            loaded_marks = json.load(f)
            for i, marks in enumerate(loaded_marks):
                clicked_x_coords[i] = set(marks)
            print(f"Loaded marks from {filename}")
    except FileNotFoundError:
        print(f"{filename} not found. Starting with empty marks.")

# Setup figure and axes
fig, ax = plt.subplots(figsize=(8, 4))
vline_hover = ax.axvline(x=0, color='gray', linestyle='--', lw=1, visible=False)
current_lines = []

# Widget to track index
index = widgets.IntText(value=0, layout=widgets.Layout(width='50px'))

# Navigation and save buttons
left_button = widgets.Button(description='←', layout=widgets.Layout(width='50px'))
right_button = widgets.Button(description='→', layout=widgets.Layout(width='50px'))
save_button = widgets.Button(description='💾 Save Labels', button_style='success')

# Threshold for deselect
CLICK_THRESHOLD = 3

def update_plot(i):
    clear_output(wait=True)
    ax.clear()

    data = plots[i].numpy() if hasattr(plots[i], 'numpy') else plots[i]
    t = list(range(len(data)))

    ax.plot(t, data, label="Truth", linestyle="solid", color="red")
    ax.set_title(f"Node {final_sorted_points[i]} Population Density")
    ax.set_xlim(0, len(data) - 1)
    ax.set_ylim(float(min(data)) - 0.1, float(max(data)) + 0.1)

    # Recreate the hover line in the current axes
    global vline_hover
    vline_hover = ax.axvline(x=0, color='gray', linestyle='--', lw=1, visible=False)

    current_lines.clear()
    for x_coord in sorted(clicked_x_coords[i]):
        vline = ax.axvline(x=x_coord, color='blue', linestyle='--', lw=1)
        current_lines.append(vline)

    display(controls)
    fig.canvas.draw_idle()
    
# Hover handler
def on_move(event):
    if event.inaxes == ax:
        vline_hover.set_xdata(event.xdata)
        vline_hover.set_visible(True)
        fig.canvas.draw_idle()

# Click handler
def on_click(event):
    if event.inaxes != ax:
        return

    x_clicked = round(event.xdata)
    plot_idx = index.value
    current_set = clicked_x_coords[plot_idx]

    for existing_x in current_set:
        if abs(existing_x - x_clicked) < CLICK_THRESHOLD:
            current_set.remove(existing_x)
            update_plot(plot_idx)
            return

    current_set.add(x_clicked)
    update_plot(plot_idx)

# Save to file
def save_marks_to_file(b):
    filename = "anomaly_labels.json"
    serializable = [sorted(list(s)) for s in clicked_x_coords]
    with open(filename, 'w') as f:
        json.dump(serializable, f, indent=2)
    print(f"Saved labels to {filename}")

# Navigation callbacks
def on_left_click(b):
    if index.value > 0:
        index.value -= 1

def on_right_click(b):
    if index.value < len(plots) - 1:
        index.value += 1

def on_index_change(change):
    if change['name'] == 'value' and change['new'] >= 0 and change['new'] < len(plots):
        update_plot(change['new'])

# Connect events
fig.canvas.mpl_connect('motion_notify_event', on_move)
fig.canvas.mpl_connect('button_press_event', on_click)
left_button.on_click(on_left_click)
right_button.on_click(on_right_click)
save_button.on_click(save_marks_to_file)

index.observe(on_index_change)

# Layout
controls = widgets.HBox([left_button, index, right_button, save_button])
display(controls)

load_marks_from_file()

# Initial plot
update_plot(index.value)
plt.show()

```

## Compute Multi-Step Model Performance Metrics

```python
def multi_step_metrics(tensor_sorted_indices):
    n_batches = y_pred.shape[0]

    y_pred_reshaped = y_pred.view(n_batches, config["BATCH_SIZE"], config["N_NODES"], config["N_PRED"])
    y_pred_top_n_9 = y_pred_reshaped[:, :, tensor_sorted_indices, :].view(n_batches, -1, config["N_PRED"])
    y_pred_top_n_6 = y_pred_top_n_9[:, :, :6]
    y_pred_top_n_3 = y_pred_top_n_9[:, :, :3]

    y_truth_reshaped = y_truth.view(n_batches, config["BATCH_SIZE"], config["N_NODES"], config["N_PRED"])
    y_truth_top_n_9 = y_truth_reshaped[:, :, tensor_sorted_indices, :].view(n_batches, -1, config["N_PRED"])
    y_truth_top_n_6 = y_truth_top_n_9[:, :, :6]
    y_truth_top_n_3 = y_truth_top_n_9[:, :, :3]

    mae_3, rmse_3, wmape_3 = models.trainer.evaluate_precomputed_predictions(y_truth_top_n_3, y_pred_top_n_3)
    mae_6, rmse_6, wmape_6 = models.trainer.evaluate_precomputed_predictions(y_truth_top_n_6, y_pred_top_n_6)
    mae_9, rmse_9, wmape_9 = models.trainer.evaluate_precomputed_predictions(y_truth_top_n_9, y_pred_top_n_9)

    headers = ["Prediction Steps", "MAE", "RMSE", "WMAPE"]
    data = [
        ["3-Step", mae_3, rmse_3, wmape_3],
        ["6-Step", mae_6, rmse_6, wmape_6],
        ["9-Step", mae_9, rmse_9, wmape_9],
    ]
    print(f"\nOver Top-{len(tensor_sorted_indices)} Nodes")
    # Print header
    print(f"{headers[0]:<20} {headers[1]:<10} {headers[2]:<10} {headers[3]:<10}")
    print("-" * 50)  # Print a line separator

    # Print data rows
    for row in data:
        print(f"{row[0]:<20} {row[1]:<10.4f} {row[2]:<10.4f} {row[3]:<10.4f}")

multi_step_metrics(torch.tensor(final_sorted_indices.copy()))
# multi_step_metrics(torch.tensor(sorted_indices.copy()))
```

Reconstruct an adjacency matrix based on a threshold of the attention matrix.

```python
attention_root = np.sqrt(attention)
attention_mean = np.mean(attention)
attention_std = np.std(attention)
threshold = attention_mean + (attention_std * 1/4)
reconstructed_adjacency = (attention > threshold).astype(int)
reconstructed_adjacency = (reconstructed_adjacency > 0).astype(int)
visualizations.adjacency_matrix.plot_adjacency_matrix(
    reconstructed_adjacency, "./output/reconstructed_adjacency.png"
)

# Read the original graph without any hop extensions or pruning
G1 = nx.read_adjlist("dataset/clustered_G1Hops.adjlist")
# Read in the pruned 3-hop graph
G2 = nx.read_adjlist("dataset/pruned_clustered_3hop_graph.adjlist")

# Keep only the nodes in the original graph that are in the pruned 3-hop graph
common_nodes = set(G1.nodes()).intersection(G2.nodes())
G = G1.subgraph(common_nodes).copy()

node_ids = list(G.nodes())
original_adjacency = nx.to_numpy_array(G, nodelist=node_ids)
original_adjacency = (original_adjacency).astype(int)

# Count total matching edges (where both have 1s)
matching_edges = np.sum((original_adjacency == 1) & (reconstructed_adjacency == 1))

# Total predicted and actual edges
total_true = np.sum(original_adjacency)
total_pred = np.sum(reconstructed_adjacency)

# Since each edge is counted twice in undirected graphs, divide by 2
matching_edges //= 2
total_true //= 2
total_pred //= 2

# Precision and recall
precision = matching_edges / total_pred if total_pred > 0 else 0
recall = matching_edges / total_true if total_true > 0 else 0

print(f"Correctly Predicted Edges: {matching_edges}")
print(f"Total Predicted Edges: {total_pred}")
print(f"Total True Edges: {total_true}")
print(f"Precision: {precision:.2%}")
print(f"Recall: {recall:.2%}")
```
